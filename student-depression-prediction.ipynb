### What Standard Deviation Actually Tells You:
It's a measure of spread: how far individual values are from the mean.

A higher standard deviation = responses are more spread out

A lower standard deviation = responses are clustered tightly around the mean

```python
df_final
```

```python
## separating our desired from our other values
X = df_final.drop(columns = ["Depression"])
Y = df_final["Depression"]
```

```python
X
```

```python
Y
```

```python
## Train and Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42, stratify=Y)
```

```python
##Training a model
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import mean_absolute_error
preds_valid=model.predict(X_test)
score_valid=mean_absolute_error(y_test,preds_valid)
print("MAE: ",score_valid)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```
